THIS WORKBENCH SERVES TO SEARCH THE OPTIMAL HYPERPARAMETERS FOR THE CONVENTIONAL SUPERVISED MODELS (SUCH AS SVM, KNN) DEFINED IN `qa_ma_sqi.py` THAT PREDICTS SQ & MA

## 1- SVMs

```python
%reset -f
import hyperparameter_optimization_utils as ut
import numpy as np

LABEL = "ma"
SIGNAL = "PPG"
CLASS = "HR"
# Use this class only for data preperation:
FEATURES = ut.get_features(LABEL, SIGNAL, CLASS)
X_train, y_train = ut.get_train_dataset(SIGNAL, LABEL, FEATURES)

# --------- #

from sklearn.svm import SVC

distributions_svm = {
    "C": np.arange(1, 10),
    "kernel": ["linear", "poly", "rbf", "sigmoid"],
    "gamma": np.arange(1, 10),
}
svc = SVC()

hp_svm = ut.search_hyper_parameters(model=svc, distributions=distributions_svm, X_train=X_train, y_train=y_train)
ut.save_hp_df(df_hp=hp_svm, model_name="svm", signal=SIGNAL, label=LABEL, class_name=CLASS)
# hp_svm = ut.load_hp_df(model_name="svm", signal=SIGNAL, label=LABEL, class_name=CLASS)
hp_svm
```

## 2- KNNs

```python
%reset -f
import hyperparameter_optimization_utils as ut
import numpy as np

LABEL = "ma"
SIGNAL = "PPG"
CLASS = "HR"
# Use this class only for data preperation:
FEATURES = ut.get_features(LABEL, SIGNAL, CLASS)
X_train, y_train = ut.get_train_dataset(SIGNAL, LABEL, FEATURES)

# --------- #

from sklearn.neighbors import KNeighborsClassifier

distributions_knn = {
    "n_neighbors": np.arange(1, 2),  # Number of neighbors from 1 to 30
    # Weight function used in prediction
    "weights": ["uniform", "distance"],
    # Algorithm used to compute the nearest neighbors
    "algorithm": ["auto", "ball_tree", "kd_tree", "brute"],
    # Power parameter for the Minkowski metric (1 is equivalent to using Manhattan distance, and 2 is equivalent to using Euclidean distance)
    "p": [1, 2],
}

knn = KNeighborsClassifier()

hp_knn = ut.search_hyper_parameters(model=knn, distributions=distributions_knn, X_train=X_train, y_train=y_train)
ut.save_hp_df(df_hp=hp_knn, model_name="knn", signal=SIGNAL, label=LABEL, class_name=CLASS)
# hp_knn = ut.load_hp_df(model_name="knn", signal=SIGNAL, label=LABEL, class_name=CLASS)
hp_knn
```
