# 0: MLP

```python
%reset -f
from ma.qa_ma_utils import load_train_test_datasets
from ma.qa_ma_networks import MLP
SIGNAL = "SCG"
LABEL = "sq" # ma or sq
CLASS_NAME = "RR"

datasets = load_train_test_datasets(signal_type=SIGNAL, label = LABEL, class_name=CLASS_NAME)
```

```python
# Initialize the network:
mlp = MLP(datasets=datasets, label = LABEL, depth = 5, activations = ["relu"]*5)
```

```python
# Train the network
history = mlp.train_network(epochs = 100, batch_size = 32)
```

```python
# Evaluate the network
eval = mlp.evaluate_network()
# Predict Data:
predictions = mlp.predict(X = mlp.X_test, ground_truth= mlp.y_test)
print(predictions)
# Save the Network:
# mlp.save_network(f"mlp_{LABEL}_{SIGNAL}_HR_1")
```
